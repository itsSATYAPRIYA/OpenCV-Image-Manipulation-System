## Take File location from user , then ask the user if he wants to do ## Rotation & flip, adding text, cropping, resizing & scaling, drawing shapes, then run each process according to user's answer and at last ask user for the output file name of the result and save the file

```python
import cv2

print("Welcome to the Open CV Image Manupulation system.")
name = input("Please Enter your name to start the program....")
print(f"Hello {name} welcome to the system. Plese read options carefully to run the processes.")

image_path = input("Enter the Path of the Image on which you want to perform the operations---")
image = cv2.imread(image_path)

if image is not None:
    height, width, channel = image.shape
    print("Image Loaded...")
    print(f"Image Informations.... \nHeight : {height}\nWidth : {width}\nChannel : {channel}")

    print("Enter your choice from the following options. (1 - 8)")
    print("1. Resize an Image")
    print("2. Crop Image")
    print("3. Image Rotation")
    print("4. Image Flipping")
    print("5. Draw Line")
    print("6. Draw Rectangle")
    print("7. Draw Circle")
    print("8. Add Text to Image")

    choice = int(input("Enter your choice: "))

    match choice:

        case 1:
            print("Image Resizing Operation.")
            resized_image = cv2.resize(image, (400, 400))

            cv2.imshow("Original Image", image)
            cv2.imshow("Resized Image", resized_image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

            output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
            cv2.imwrite(output_file_name, resized_image)
            print(f"File saved as {output_file_name} successfully.")

        case 2:
            print("Image Cropping Operation.")
            y1, y2 = map(int, input("Enter the values of y1 & y2 accordingly...").split())
            x1, x2 = map(int, input("Enter the value of x1 & x2 accordingly...").split())

            croppped_image = image[y1:y2, x1:x2]

            cv2.imshow("Original Image.", image)
            cv2.imshow("Cropped Image", croppped_image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

            output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
            cv2.imwrite(output_file_name, croppped_image)
            print(f"File saved successfully {output_file_name} successfully.")

        case 3:
            print("Image Rotation Operation.")
            angle = int(input("Enter the angle on which the image will be rotated :"))
            (h, w) = image.shape[:2]

            center = (w // 2, h // 2)
            Matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
            rotated_image = cv2.warpAffine(image, Matrix, (w, h))

            cv2.imshow("Original", image)
            cv2.imshow(f"Rotated {angle} degree", rotated_image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

            output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
            cv2.imwrite(output_file_name, rotated_image)
            print(f"File saved as {output_file_name} successfully.")

        case 4:
            print("Image Flip Operation.")
            print("Enter how you want to flip the image ? (1 - 3)")
            print("1.Horizontal Image flip\n2.Vertical Image flip\n3.Both Vertically & Horizontally")

            option = int(input("Enter your choice : "))

            if option == 1:
                horizontal_flip = cv2.flip(image, 1)

                cv2.imshow("Horizontally Flipped Image", horizontal_flip)
                cv2.waitKey(0)
                cv2.destroyAllWindows()

                output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
                cv2.imwrite(output_file_name, horizontal_flip)
                print(f"File saved as {output_file_name} successfully.")

            elif option == 2:
                vertical_flip = cv2.flip(image, 0)

                cv2.imshow("Vertically Flipped Image", vertical_flip)
                cv2.waitKey(0)
                cv2.destroyAllWindows()

                output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
                cv2.imwrite(output_file_name, vertical_flip)
                print(f"File saved as {output_file_name} successfully.")

            elif option == 3:
                both_flip = cv2.flip(image, -1)

                cv2.imshow("Both Horizontally & Vertically", both_flip)
                cv2.waitKey(0)
                cv2.destroyAllWindows()

                output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
                cv2.imwrite(output_file_name, both_flip)
                print(f"File saved as {output_file_name} successfully.")

            else:
                print("Wrong choice Entered !")

        case 5:
            print("Line Drawing on image Operation.")
            x1, y1 = map(int, input("Enter the values of x1 & y1 accordingly : ").split())
            x2, y2 = map(int, input("Enter the values of x2, y2 accordingly : ").split())
            thickness = int(input("Enter the thickness of the line : "))
            color = (255, 255, 0)

            cv2.line(image, (x1, y1), (x2, y2), color, thickness)
            cv2.imshow("Line on the image", image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

        case 6:
            print("Rectangle Drawing on Image Operation.")
            x1, y1 = map(int, input("Enter the values of x1 & y1 accordingly : ").split())
            x2, y2 = map(int, input("Enter the values of x2, y2 accordingly : ").split())
            thickness = int(input("Enter the thickness of the line : "))
            color = (255, 255, 0)

            cv2.rectangle(image, (x1, y1), (x2, y2), color, thickness)
            cv2.imshow("Rectangle on the image", image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

        case 7:
            print("Circle Drawing on Image Operation.")
            x, y = map(int, input("Enter the values of x & y for circle centre: ").split())
            radius = int(input("Enter the radius of the circle"))
            thickness = int(input("Enter the thickness of the circle"))
            color = (255, 255, 0)

            cv2.circle(image, (x, y), radius, color, thickness)
            cv2.imshow("Circle drawing", image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

        case 8:
            print("Add text on Image Operation.")
            text_input = input("Enter the text you want to display on the image : ")
            x, y = map(int, input("Enter the values of x & y for the postion of text accordingly : ").split())
            color = (255, 255, 78)
            thickness = int(input("Enter the thickness of the text which will be shown on the image : "))

            new_image = cv2.putText(
                image,
                text_input,
                (x, y),
                cv2.FONT_HERSHEY_SIMPLEX,
                2,
                color,
                thickness
            )

            cv2.imshow(text_input, new_image)
            cv2.waitKey(0)
            cv2.destroyAllWindows()

            output_file_name = input("Enter the output file name...(e.g. output.png or output.jpg)")
            cv2.imwrite(output_file_name, new_image)
            print(f"File saved as {output_file_name} successfully.")

        case _:
            print("Invalid Choice Entered! Process terminated, please start again. Thank You")
```
